```python
# ==============================
# OTT TREND ANALYSIS PROJECT
# ==============================

# Import Libraries

import pandas as pd
import numpy as np

# Visualization Libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries Imported Successfully")
```

```python
# Load Dataset

df = pd.read_csv('tmdb_5000_movies.csv')

print("Dataset Loaded Successfully")
```

```python
# Display first 5 rows

df.head()
```

```python
# Dataset Information

df.info()
```

```python
# Check Missing Values

df.isnull().sum()
```

```python
# Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print("Duplicate Rows Removed")
```

```python
# Convert all column names to lowercase
df.columns = df.columns.str.lower()

# Remove extra spaces
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# Fill Missing Values

df['homepage'].fillna('Not Available', inplace=True)
df['overview'].fillna('No Overview', inplace=True)
df['tagline'].fillna('No Tagline', inplace=True)

print("Missing Values Handled")
```

```python
# Convert release_date column

df['release_date'] = pd.to_datetime(df['release_date'])

print("Date Converted Successfully")
```

```python
# Extract Release Year

df['release_year'] = df['release_date'].dt.year

# Extract Release Month

df['release_month'] = df['release_date'].dt.month

print("New Features Created")
```

```python
# Top Movies by Popularity

df[['title', 'popularity']].sort_values(
    by='popularity',
    ascending=False
).head(10)
```

```python
# Top 10 Popular Movies

top_movies = df[['title', 'popularity']] \
                .sort_values(by='popularity', ascending=False) \
                .head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x='popularity',
    y='title',
    data=top_movies
)

plt.title("Top 10 Most Popular Movies")

plt.show()
```

```python
# Movies Released Per Year

release_trend = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))

release_trend.plot()

plt.title("Movies Released Over Years")

plt.xlabel("Year")
plt.ylabel("Number of Movies")

plt.show()
```

```python
# Top Rated Movies

top_rated = df[['title', 'vote_average']] \
                .sort_values(by='vote_average', ascending=False) \
                .head(10)

top_rated
```

```python
# Visualization of Top Rated Movies

plt.figure(figsize=(12,6))

sns.barplot(
    x='vote_average',
    y='title',
    data=top_rated
)

plt.title("Top Rated Movies")

plt.show()
```

```python
# Top Revenue Movies

top_revenue = df[['title', 'revenue']] \
                .sort_values(by='revenue', ascending=False) \
                .head(10)

top_revenue
```

```python
# Revenue Visualization

plt.figure(figsize=(12,6))

sns.barplot(
    x='revenue',
    y='title',
    data=top_revenue
)

plt.title("Top Revenue Movies")

plt.show()
```

```python
# Runtime Distribution

plt.figure(figsize=(10,6))

sns.histplot(df['runtime'], bins=30)

plt.title("Movie Runtime Distribution")

plt.xlabel("Runtime (minutes)")

plt.show()
```

```python
# Top Languages

top_languages = df['original_language'].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_languages.index,
    y=top_languages.values
)

plt.title("Top Movie Languages")

plt.show()
```

```python
# Business Insights

most_popular = df.loc[df['popularity'].idxmax(), 'title']

highest_revenue = df.loc[df['revenue'].idxmax(), 'title']

highest_rated = df.loc[df['vote_average'].idxmax(), 'title']

average_runtime = df['runtime'].mean()

print("===== BUSINESS INSIGHTS =====")

print(f"Most Popular Movie : {most_popular}")

print(f"Highest Revenue Movie : {highest_revenue}")

print(f"Highest Rated Movie : {highest_rated}")

print(f"Average Runtime : {average_runtime:.2f} minutes")
```

```python
# Save Cleaned Dataset

df.to_csv('cleaned_ott_movies_dataset.csv', index=False)

print("Dataset Saved Successfully")
```
