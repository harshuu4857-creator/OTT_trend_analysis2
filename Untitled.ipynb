```python
# Function to extract genre names

def extract_genres(text):
    
    genres_list = []
    
    try:
        data = ast.literal_eval(text)
        
        for item in data:
            genres_list.append(item['name'])
            
        return ', '.join(genres_list)
    
    except:
        return 'Unknown'
```

```python
# Function to extract production companies

def extract_companies(text):
    
    companies = []
    
    try:
        data = ast.literal_eval(text)
        
        for item in data:
            companies.append(item['name'])
            
        return ', '.join(companies)
    
    except:
        return np.nan
```

```python
# Create cleaned production companies column

df['production_companies_cleaned'] = df['production_companies'].apply(extract_companies)

print("Production Companies Cleaned")
```

```python
df[['production_companies', 'production_companies_cleaned']].head()
```

```python
# Split companies

all_companies = df['production_companies_cleaned'] \
                    .str.split(', ') \
                    .explode()

# Top companies

top_companies = all_companies.value_counts().head(10)

top_companies
```

```python
# Top Production Companies Chart

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_companies.values,
    y=top_companies.index
)

plt.title("Top Production Companies")

plt.xlabel("Number of Movies")
plt.ylabel("Production Company")

plt.show()
```
